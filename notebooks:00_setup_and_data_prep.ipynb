## ۳) لیست اسلات‌ها

```python
import sys
sys.path.append(os.path.abspath("."))
from chatbot_common import canonical_slot_name

RAW_SLOTS_TXT = os.path.join(DATA_DIR, "slots.txt")

with open(RAW_SLOTS_TXT, encoding="utf-8") as f:
    bio_labels = [line.strip() for line in f if line.strip()]

canonical = sorted({canonical_slot_name(l) for l in bio_labels if l != "o"})

slots_payload = {
    "bio_labels": bio_labels,     
    "canonical": canonical,     
}
with open(os.path.join(DATA_DIR, "slots.json"), "w", encoding="utf-8") as f:
    json.dump(slots_payload, f, ensure_ascii=False, indent=2)

print(f"{len(bio_labels)} برچسب BIO -> {len(canonical)} اسلات یکتا. ذخیره شد در data/slots.json")
```

## ۵)  ساخت نمونه‌های Few-shot از داده‌ی آموزشی

```python
from chatbot_common import load_raw_examples, bio_slots_to_dict

TRAIN_DIR = os.path.join(DATA_DIR, "train")
N_PER_INTENT = 2

raw_train = load_raw_examples(TRAIN_DIR)
print(f"{len(raw_train)} نمونه‌ی آموزشی یافت شد در {TRAIN_DIR}")

few_shot = {}
if raw_train:
    by_intent = {}
    intent_code_to_name = {}
    if os.path.exists(INTENT_MAPPING_CSV):
        mdf = pd.read_csv(INTENT_MAPPING_CSV)
        intent_code_to_name = dict(zip(mdf["code"], mdf["intent"]))

    for ex in raw_train:
        intent_name = intent_code_to_name.get(ex.get("intent_id"), ex.get("intent_id"))
        by_intent.setdefault(intent_name, []).append(ex)

    for intent_name, examples in by_intent.items():
        sample = random.sample(examples, min(N_PER_INTENT, len(examples)))
        few_shot[intent_name] = [
            {
                "text": ex["input_text"],
                "slots": bio_slots_to_dict(ex["input_text"], ex.get("slots", [])),
            }
            for ex in sample
        ]

    with open(os.path.join(DATA_DIR, "few_shot_examples.json"), "w", encoding="utf-8") as f:
        json.dump(few_shot, f, ensure_ascii=False, indent=2)
    print(f"few-shot examples برای {len(few_shot)} intent ذخیره شد -> data/few_shot_examples.json")
else:
    print("داده‌ی آموزشی پیدا نشد).")
```
